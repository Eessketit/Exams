### **Put all together**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Data
x = np.array(range(0, 100))
y = 32 + 1.8 * x

data = {
    "input (x)": x,
    "target (y)": y
}
df = pd.DataFrame(data=data)

# Parameters
m = len(x)
theta_0 = 0
theta_1 = 0
# theta_0 = np.random.rand()
# theta_1 = np.random.rand()
lr = 0.0001

# Gradient descent
iterations = 1000000
costs = []

for i in range(iterations):
    # Prediction
    h_x = theta_0 + theta_1 * x

    # Cost function
    cost = (1/(2*m)) * np.sum((h_x - y)**2)
    costs.append(cost)

    # Gradients
    d_theta0 = (1/m) * np.sum(h_x - y)
    d_theta1 = (1/m) * np.sum((h_x - y) * x)

    # Update parameters
    theta_0 = theta_0 - lr * d_theta0
    theta_1 = theta_1 - lr * d_theta1

print("theta_0:", theta_0)
print("theta_1:", theta_1)

plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost function decrease")
plt.show()
```

## **Applied Normalization**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ---------------------------
# Data preparation
# ---------------------------
x = np.array(range(0, 100))
y = 32 + 1.8 * x


# Train-test split 80/20 ratio

x_train, x_test = x[:80], x[80:]
y_train, y_test = y[:80], y[80:]

# ---------------------------
# Normalization (Standardization)
# ---------------------------
x_mean = np.mean(x_train)
x_std = np.std(x_train)
x_train_norm = (x_train - x_mean) / x_std
x_test_norm = (x_test - x_mean) / x_std   # use train mean/std for test too

print(f"Number of samples of training set: {len(x_train_norm)}")
print(f"Number of samples of test set: {len(x_test_norm)}")
print("Test samples \n", x_test_norm)
```

```python
# ---------------------------
# Training (Gradient Descent)
# ---------------------------
m = len(x_train)
theta_0, theta_1 = 0, 0
lr = 0.01
iterations = 1000
costs = []

for i in range(iterations):
    # Prediction
    h_x = theta_0 + theta_1 * x_train_norm

    # Cost function
    cost = (1 / (2 * m)) * np.sum((h_x - y_train) ** 2)
    costs.append(cost)

    # Gradients
    d_theta0 = (1/m) * np.sum(h_x - y_train)
    d_theta1 = (1/m) * np.sum((h_x - y_train) * x_train_norm)

    # Update
    theta_0 -= lr * d_theta0
    theta_1 -= lr * d_theta1

print("Learned parameters:")
print("theta_0:", theta_0)
print("theta_1:", theta_1)
print("Cost values: ", cost)
```

```python
plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost function decrease")
plt.show()
```

```python
# ---------------------------
# Visualization
# ---------------------------
plt.scatter(x_train, y_train, color="blue", label="Training data")
y_pred_train = theta_0 + theta_1 * x_train_norm
plt.plot(x_train, y_pred_train, color="red", label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# Testing
y_pred_test = theta_0 + theta_1 * x_test_norm
mse = np.mean((y_pred_test - y_test) ** 2)

print("\nTesting Results:")
for i in range(len(x_test)):
    print(f"x={x_test[i]}, Original={y_test[i]:.2f}, Predicted={y_pred_test[i]:.2f}")

print("\nMean Squared Error on Test Data:", mse)
```

### **Regularization technique**

```python
import matplotlib.pyplot as plt
import numpy as np

dataset = np.array(([10, 12, 5, 50],
                    [15, 11, 7, 59],
                    [20, 23, 3, 68],
                    [25, 28, 6, 77],
                    [30, 31, 6, 86]))

print(dataset.shape)
X = dataset[:, 0:3]
y = dataset[:, -1]

y = y.reshape(len(y), 1)

print(X.shape)
print(y.shape)
x0 = np.ones((5, 1))
X_fin = np.hstack((x0, X))
print(X_fin.shape)
print(X_fin)

learning_rate = 0.001
iteration = 1000
regularization_param = 0.1  # Lambda for regularization term
Loss = []

theta = np.zeros((4, 1))
m = len(X_fin)

for iter in range(iteration):
    h_x = np.dot(X_fin, theta)
    error = h_x - y
    loss = (np.sum(error**2) + regularization_param * np.sum(theta[1:]**2)) / (2 * m)
    Loss.append(loss)

    # Gradient descent with regularization (excluding bias term from regularization)
    delta = (np.dot(np.transpose(X_fin), error) + regularization_param * np.vstack(([0], theta[1:]))) / m
    theta = theta - learning_rate * delta

    if iter % 100 == 0:
        print(f"{iter} iteration, Loss = {loss}")

x_axis = np.array(range(iteration))
y_axis = np.array(Loss)
plt.plot(x_axis, y_axis)
plt.xlabel("Number of iterations")
plt.ylabel("Loss")
plt.show()
```
