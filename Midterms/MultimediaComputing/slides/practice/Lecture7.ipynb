```python
def srgb_to_xyz(r, g, b):
    #  Normalize 0-255 RGB values to the range 0.0 - 1.0
    r_norm = r / 255.0
    g_norm = g / 255.0
    b_norm = b / 255.0

    # Inverse Gamma Correction (Linearize the sRGB values)
    def linearize(c):
        if c <= 0.04045:
            return c / 12.92
        else:
            return ((c + 0.055) / 1.055) ** 2.4

    r_lin = linearize(r_norm)
    g_lin = linearize(g_norm)
    b_lin = linearize(b_norm)

    # Apply the transformation matrix for sRGB to XYZ (D65 Illuminant)
    X = r_lin * 0.4124564 + g_lin * 0.3575761 + b_lin * 0.1804375
    Y = r_lin * 0.2126729 + g_lin * 0.7151522 + b_lin * 0.0721750
    Z = r_lin * 0.0193339 + g_lin * 0.1191920 + b_lin * 0.9503041

    # Calculate 2D chromaticity coordinates (x, y)
    total = X + Y + Z
    if total == 0:
        x, y = 0.3127, 0.3290  # Default to D65 white point if color is pure black
    else:
        x = X / total
        y = Y / total

    return {"X": round(X, 4), "Y": round(Y, 4), "Z": round(Z, 4), "x": round(x, 4), "y": round(y, 4)}

# --- Use different colors here ---
test_colors = {
    "Pure Red": (255, 0, 0),
    "Pure Green": (0, 255, 0),
    "Pure White": (255, 255, 255),
    "Mid Gray": (128, 128, 128)
  #  "Yellow": (255, 255, 0),         # Yellow (Red + Green)
  #  "Cyan": (0, 255, 255),           # Cyan (Green + Blue)
}

for name, rgb in test_colors.items():
    coords = srgb_to_xyz(*rgb)
    print(f"=== {name} {rgb} ===")
    print(f"XYZ Tristimulus Values: X={coords['X']}, Y={coords['Y']}, Z={coords['Z']}")
    print(f"2D Chromaticity Plane:  x={coords['x']},  y={coords['y']}\n")
```
